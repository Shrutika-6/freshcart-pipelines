```
from iris_project.data_loaders.load_iris import load_data
from iris_project.transformers.validate_iris import validate_data
from iris_project.transformers.prepare_data import transform
from iris_project.transformers.train_model import train
from iris_project.data_exporters.save_results import export_data
from iris_project.data_exporters.evaluate_model import evaluate

# 1. Load data
df = load_data()
print('Loaded iris data with shape:', df.shape)

# 2. Validate data
df_val = validate_data(df)

# 3. Prepare data
df_prep = transform(df_val, test_size=0.2)
print('Split counts:
', df_prep['split'].value_counts())

# 4. Train model (max_depth=3)
model_out = train(df_prep, max_depth=3)
print('Test Accuracy:', model_out['test_accuracy'])

# 5. Save results and Evaluate
export_data(model_out)
eval_out = evaluate(model_out)
```
